### Мета
Реалізувати та дослідити роботу бінарного дерева пошуку (BST), бінарної купи (Heap) та хеш-таблиці.

## Виконання
1. **Дерево (BST):** структура, де кожен вузол має лівого нащадка (менше значення) і правого (більше значення).
2. **Купа (Heap):** деревоподібна структура, яка підтримує властивість "найбільший/найменший елемент у корені".
3. **Хеш-таблиця:** структура для пошуку за ключем з середньою складністю $O(1)$.

```python
# 1. Хеш-таблиця (через словник Python)
hash_table = {"Ключ1": 100, "Ключ2": 200}
print("Хеш-таблиця:", hash_table)

# 2. Купа (Heap) через heapq
import heapq
heap = [5, 1, 8, 3]
heapq.heapify(heap) # Перетворюємо список на купу
print("Купа (heap):", heap)

# 3. Бінарне дерево пошуку (проста реалізація вузла)
class Node:
    def __init__(self, key):
        self.val = key
        self.left = None
        self.right = None

root = Node(10)
root.left = Node(5)
root.right = Node(15)
print(f"Корінь дерева: {root.val}, лівий нащадок: {root.left.val}")
```

## Висновок
Під час роботи я реалізував три фундаментальні структури даних. Хеш-таблиці забезпечують найшвидший доступ до даних за ключем. Купи є ідеальними для реалізації черг з пріоритетом. Бінарні дерева пошуку дозволяють ефективно зберігати впорядковані дані з логарифмічною складністю пошуку $O(\log n)$.

## Контрольні питання

### 1. Чим відрізняється Купа від Дерева пошуку?
У купи властивість лише в тому, що корінь має екстремальне значення, тоді як у BST всі ліві елементи менші за вузол, а праві — більші.

### 2. Як працює хеш-функція?
Вона перетворює вхідний ключ будь-якої довжини на число (індекс) фіксованої довжини, вказуючи на комірку в пам'яті (кошик).

### 3. Що таке колізія в хеш-таблиці?
Це ситуація, коли два різних ключі мають однаковий хеш-код. Вирішується методами ланцюжків (списки) або відкритої адресації.
