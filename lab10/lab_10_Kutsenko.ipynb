### Виконання
Кодування Гафмена — це алгоритм стиснення даних без втрат (жадібна стратегія). Ідея полягає в тому, що часто вживаним символам присвоюються короткі коди, а рідко вживаним — довгі. Використовуються префіксні коди — жоден код не є початком іншого, що гарантує відсутність неоднозначності при зворотному декодуванні. 
Для ефективної побудови дерева використовується структура даних «Купа» (Heap), що діє як черга з пріоритетом.

```python
import heapq
from collections import Counter

# 1. Вузол дерева Гафмена (з виправленим магічним методом __lt__ для heapq)
class node:
    def __init__(self, freq, symbol, left=None, right=None):
        self.freq = freq
        self.symbol = symbol
        self.left = left
        self.right = right
        self.huff = ''

    def __lt__(self, nxt):
        return self.freq < nxt.freq

# Словник для збереження готових кодів
huffman_codes = {}

# Утиліта для виведення та збереження кодів Гафмена
def printNodes(node, val=''):
    newVal = val + str(node.huff)
    if node.left:
        printNodes(node.left, newVal)
    if node.right:
        printNodes(node.right, newVal)
    if not node.left and not node.right:
        print(f"'{node.symbol}' -> {newVal}")
        huffman_codes[node.symbol] = newVal

# --- ВИКОНАННЯ ЗАВДАНЬ ДЛЯ САМОСТІЙНОЇ РОБОТИ ---

# Вхідне повідомлення
message = "алгоритми_та_методи_обчислень"
print(f"Оригінальне повідомлення: {message}\n")

# Завдання 1: Процедура обчислення частот
freq_dict = Counter(message)
chars = list(freq_dict.keys())
freq = list(freq_dict.values())

print("Символи:", chars)
print("Частоти:", freq)

# Побудова дерева
nodes = []
for x in range(len(chars)):
    heapq.heappush(nodes, node(freq[x], chars[x]))

while len(nodes) > 1:
    left = heapq.heappop(nodes)
    right = heapq.heappop(nodes)
    left.huff = 0
    right.huff = 1
    # Створюємо новий внутрішній вузол
    newNode = node(left.freq + right.freq, left.symbol + right.symbol, left, right)
    heapq.heappush(nodes, newNode)

root = nodes[0]
print("\nЗгенеровані коди Гафмена:")
printNodes(root)

# Завдання 2: Декодування Гафмена
def decode_huffman(encoded_str, root):
    decoded_str = ""
    current_node = root
    for bit in encoded_str:
        if bit == '0':
            current_node = current_node.left
        else:
            current_node = current_node.right
        
        # Якщо дійшли до листка (символу)
        if current_node.left is None and current_node.right is None:
            decoded_str += current_node.symbol
            current_node = root # Повертаємось у корінь для наступного символу
    return decoded_str

# Тестування всього процесу
encoded_message = "".join([huffman_codes[char] for char in message])
print(f"\nЗакодоване повідомлення (біти): \n{encoded_message}")

decoded_message = decode_huffman(encoded_message, root)
print(f"\nДекодоване повідомлення: \n{decoded_message}")
print(f"Чи збігається результат з оригіналом? {'Так' if message == decoded_message else 'Ні'}")
```

Використання структури даних «Купа» (реалізованої через модуль `heapq`) дозволило ефективно отримувати вузли з мінімальною частотою, забезпечуючи загальну асимптотичну складність $O(n \log n)$. Написаний алгоритм декодування успішно і без втрат відновив початкове повідомлення.

## Контрольні питання

### 1. Що таке жадібні алгоритми?
Жадібні алгоритми — це клас алгоритмів, які на кожному кроці роблять локально оптимальний вибір з надією, що ця послідовність виборів призведе до глобально оптимального розв'язку всієї задачі. Кодування Гафмена є успішним прикладом, де такий підхід дає ідеальний результат.

### 2. Що таке префіксний код? Який код використовується у коді Гафмена?
Префіксний код — це метод кодування, при якому жодне кодове слово (послідовність бітів) не є початком (префіксом) іншого кодового слова. Саме цей тип кодів використовується в алгоритмі Гафмена, що дозволяє декодувати суцільний потік бітів без використання роздільників між символами.

### 3. Як пов'язана структура даних «купа» зі структурою даних «черга з пріоритетами»?
Черга з пріоритетами — це абстрактний тип даних, з якого елементи витягуються в порядку їх важливості (пріоритету). «Купа» (Heap) — це конкретна структура даних (зазвичай у вигляді бінарного дерева), яка є найефективнішим способом реалізації черги з пріоритетами, дозволяючи додавати та вилучати елементи за час $O(\log n)$.
