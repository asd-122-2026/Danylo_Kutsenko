## Хід роботи
### Завдання
Реалізувати рекурсивний алгоритм (наприклад, обчислення факторіала або чисел Фібоначчі), провести оцінку його часової складності та перевірити результати на практиці.

### Виконання
Рекурсивні алгоритми базуються на самовиклику функції. Згідно з основною теоремою рекурсії, для алгоритмів вигляду $T(n) = aT(n/b) + f(n)$ ми можемо визначити асимптотичну складність.

```python
import time
import matplotlib.pyplot as plt

# Рекурсивний алгоритм обчислення факторіала
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial_recursive(n - 1)

# Дослідження часу виконання
sizes = [10, 100, 200, 300, 400, 500, 600, 700, 800, 900]
times = []

for n in sizes:
    start = time.time()
    factorial_recursive(n)
    times.append(time.time() - start)

# Графік
plt.plot(sizes, times, marker='o', color='red')
plt.xlabel("Розмір входу (n)")
plt.ylabel("Час (секунди)")
plt.title("Часова складність рекурсивного факторіала")
plt.grid(True)
plt.show()
```

## Висновок
У ході лабораторної роботи я реалізував рекурсивний алгоритм та дослідив його часову складність. Рекурсія дозволяє спростити код, проте важливо стежити за глибиною стеку, оскільки надмірна кількість викликів може призвести до помилки переповнення (RecursionError).

## Контрольні питання

### 1. Що таке основна теорема рекурсії?
Це метод для розв'язання рекурентних співвідношень вигляду $T(n) = aT(n/b) + f(n)$, який дозволяє швидко знайти асимптотичну складність алгоритму (наприклад, $O(n \log n)$ або $O(n^k)$).

### 2. Чим небезпечна неконтрольована рекурсія?
Вона може спричинити вичерпання пам'яті стеку (Stack Overflow), оскільки кожен виклик функції зберігає свій стан у пам'яті.

### 3. Чи можна перетворити будь-яку рекурсію на ітерацію?
Так, будь-який рекурсивний алгоритм можна переписати в ітеративний (через цикли), використовуючи власну структуру даних — стек.
