## Хід роботи
### Завдання
Реалізувати представлення графа в пам'яті за допомогою списку суміжності. Написати та протестувати алгоритми обходу графа: пошук у ширину (BFS) та пошук у глибину (DFS).

### Виконання
Граф — це абстрактна структура даних, що складається з множини вершин та ребер. Для обходу графа найчастіше використовують два алгоритми: BFS (досліджує граф по рівнях) та DFS (досліджує граф максимально глибоко вздовж кожної гілки).

```python
from collections import deque

# --- 1. Представлення графа (Список суміжності) ---
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# --- 2. Алгоритм пошуку в ширину (BFS) ---
def bfs(graph_data, start_vertex):
    visited = set()
    queue = deque([start_vertex])
    visited.add(start_vertex)
    traversal_order = []

    while queue:
        vertex = queue.popleft()
        traversal_order.append(vertex)

        for neighbor in graph_data[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
                
    return traversal_order

# --- 3. Алгоритм пошуку в глибину (DFS) ---
def dfs(graph_data, start_vertex, visited=None, traversal_order=None):
    if visited is None:
        visited = set()
        traversal_order = []

    visited.add(start_vertex)
    traversal_order.append(start_vertex)

    for neighbor in graph_data[start_vertex]:
        if neighbor not in visited:
            dfs(graph_data, neighbor, visited, traversal_order)
            
    return traversal_order

# --- 4. Демонстрація роботи ---
print("Представлення графа (словник):")
for node, edges in graph.items():
    print(f"  Вершина {node} з'єднана з {edges}")

print("\nОбхід графа в ширину (BFS) з вершини 'A':")
print(" -> ".join(bfs(graph, 'A')))

print("\nОбхід графа в глибину (DFS) з вершини 'A':")
print(" -> ".join(dfs(graph, 'A')))
```

Також я на практиці дослідив роботу базових алгоритмів обходу — пошуку в ширину (BFS) за допомогою черги та пошуку в глибину (DFS) з використанням рекурсії.

## Контрольні питання

### 1. Які існують способи представлення графа в пам'яті? Які їхні переваги та недоліки?
Основні способи — це матриця суміжності, список суміжності та матриця інцидентності. Матриця суміжності забезпечує швидкий доступ до інформації про наявність ребра $O(1)$, але вимагає багато пам'яті $O(V^2)$, що робить її ефективною лише для щільних графів. Список суміжності більш економний до пам'яті і є оптимальним для розріджених графів.

### 2. Як працює алгоритм BFS?
Алгоритм пошуку в ширину (BFS) обходить граф "рівнями". Починаючи з початкової вершини, він спочатку відвідує всіх її безпосередніх сусідів, потім — сусідів сусідів і так далі. Для збереження черговості вершин використовується структура даних Черга (Queue). Застосовується для знаходження найкоротшого шляху у незважених графах та аналізу мереж.

### 3. Що таке DFS і чим він відрізняється від BFS?
DFS (пошук у глибину) — це алгоритм, який рухається по гілці графа максимально глибоко, поки не досягне "глухого кута", після чого повертається назад (backtracking). Основна відмінність: DFS використовує стек або рекурсію, тоді як BFS використовує чергу. DFS чудово підходить для пошуку циклів у графі та топологічного сортування.

### 4. Опишіть алгоритм Дейкстри.
Алгоритм Дейкстри використовується для знаходження найкоротших шляхів від однієї заданої вершини до всіх інших вершин у зваженому графі. Головна умова його коректної роботи — ваги всіх ребер графа повинні бути невід'ємними числами.
