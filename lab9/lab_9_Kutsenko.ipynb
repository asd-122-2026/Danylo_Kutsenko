### Виконання
[cite_start]Рядок — це послідовність символів із заданої множини символів[cite: 825]. Для роботи з ними застосовуються різні підходи:
1. [cite_start]**Наївний алгоритм** перевіряє, чи всі символи основного рядка збігаються з шаблоном, послідовно зсуваючи його[cite: 830, 831].
2. [cite_start]**Z-функція** від рядка $s$ довжини $n$ — це масив, $i$-ий елемент якого дорівнює найбільшому числу символів, починаючи з позиції $i$, які співпадають із першими символами рядка $s$[cite: 868]. [cite_start]Вона є більш ефективною для обробки великих текстів[cite: 866].

```python
# 1. Наївний пошук шаблонів
def naive_match(p, t):
    assert len(p) <= len(t) # припускаємо, що довжина шаблону не більша за довжину тексту
    occurrences = []
    for i in range(0, len(t) - len(p) + 1):
        match = True
        for j in range(0, len(p)): # для кожної позиції шаблону
            if t[i+j] != p[j]:
                match = False # принаймні один символ не збігається
                break
        if match:
            occurrences.append(i)
    return occurrences

# 2. Обчислення Z-функції
def z_func(s):
    # Ініціалізуємо масив Z, в якому будемо зберігати значення Z-функції
    Z = [len(s)] + [0] * len(s)
    assert len(s) > 1 
    
    # Початкове порівняння s[1:] з префіксом
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            Z[1] += 1
        else:
            break
            
    r, l = 0, 0
    if Z[1] > 0:
        r, l = Z[1], 1
        
    for k in range(2, len(s)):
        assert Z[k] == 0 
        if k > r:
            for i in range(k, len(s)):
                if s[i] == s[i-k]:
                    Z[k] += 1
                else:
                    break
            if Z[k] > 0:
                r, l = k + Z[k] - 1, k
        else:
            nbeta = r - k + 1
            zkp = Z[k - l]
            if nbeta > zkp:
                Z[k] = zkp
            else:
                nmatch = 0
                for i in range(r + 1, len(s)):
                    if s[i] == s[i-k]:
                        nmatch += 1
                    else:
                        break
                l, r = k, r + nmatch
                Z[k] = r - k + 1
    # Повертаємо масив коректної довжини
    return Z[:len(s)]

# 3. Пошук підрядка за допомогою Z-функції
def zMatch(p, t):
    s = p + "$" + t
    z = z_func(s)
    occurrences = []
    for i in range(len(p) + 1, len(s)):
        if z[i] == len(p):
            occurrences.append(i - (len(p) + 1))
    return occurrences

# 4. Стиснення рядка
def compress_with_z(s):
    z_vec = z_func(s)
    for i in range(1, len(s)):
        if (i + z_vec[i] == len(s)) and (z_vec[i] % i == 0):
            return s[:i]
    return s

# ==========================================
# ТЕСТУВАННЯ (Завдання для самостійної роботи)
# ==========================================
print("--- 1. Наївний пошук ---")
print("Шукаємо 'needle' у 'needleneedleneedle':", naive_match('needle', 'needleneedleneedle'))

print("\n--- 2. Z-функція ---")
print("Z-функція для 'abracadabra':", z_func('abracadabra'))

print("\n--- 3. Пошук за допомогою Z-функції ---")
print("Шукаємо 'аб' у 'абабагаламага':", zMatch('аб', 'абабагаламага'))

print("\n--- 4. Стиснення рядка ---")
s_test = "абырвалгабырвалгабырвалг"
print(f"Оригінальний рядок: {s_test}")
print(f"Стиснутий рядок:    {compress_with_z(s_test)}")
```

## Завдання для самостійної роботи

### [cite_start]1. Асимптотика наївного алгоритму пошуку підрядка [cite: 1038]
У найгіршому випадку алгоритм порівнює кожен символ шаблону $p$ з кожним символом тексту $t$ для кожної можливої позиції зсуву. [cite_start]Тому асимптотична складність у найгіршому випадку становить $O(|t| \cdot |p|)$[cite: 840]. [cite_start]Це робить алгоритм неоптимальним для великих текстів[cite: 865].

### [cite_start]2. Асимптотична складність алгоритму стиснення рядка за допомогою Z-функції [cite: 1039, 1040]
Алгоритм складається з двох етапів: спочатку обчислюється Z-функція, що займає лінійний час $O(n)$. Потім виконується один прохід по масиву `z_vec` в циклі від $1$ до $n$, що також займає $O(n)$ операцій. Отже, загальна асимптотична складність алгоритму стиснення становить $O(n)$.

## Контрольні питання

### [cite_start]1. Що таке «префікс-функція» у контексті алгоритмів на рядках, і як вона відрізняється від Z-функції? [cite: 1061, 1062]
Префікс-функція (використовується в алгоритмі Кнута-Морріса-Пратта) для кожної позиції зберігає довжину найбільшого власного префікса, який одночасно є суфіксом підрядка, що закінчується в цій позиції. [cite_start]Z-функція ж зберігає найбільшу кількість символів, починаючи з позиції $i$, які збігаються з початком (префіксом) усього рядка[cite: 868]. 

### [cite_start]2. Що таке Z-функція у вищому контексті алгоритмів на рядках, і яка її роль у вирішенні задач? [cite: 1063]
Z-функція — це потужний інструмент лінійного часу, який дозволяє знаходити закономірності (патерни) у рядку. Її головна роль полягає в оптимізації задач пошуку. [cite_start]Вона застосовується для ефективного пошуку підрядка у рядку, знаходження кількості різних підрядків та визначення найкоротшого періоду для стиснення рядка[cite: 955, 956, 957].

### [cite_start]3. Які існують підходи до вирішення задачі «найдовший спільний підрядок» для двох рядків? [cite: 1064]
Цю задачу можна вирішити кількома методами:
* Використання двовимірного масиву (Динамічне програмування) зі складністю $O(n \cdot m)$.
* Побудова Суфіксного дерева (Suffix Tree) або Суфіксного масиву (Suffix Array) для швидкого пошуку за $O(n + m)$.

### [cite_start]4. Як можна застосувати алгоритми на рядках у задачах обробки природної мови або обробки текстів? [cite: 1065, 1066]
Алгоритми на рядках є фундаментом для:
* Токенізації тексту (розбиття на слова або речення).
* [cite_start]Пошукових систем та баз даних (знаходження ключових слів у великих масивах даних)[cite: 827].
* Перевірки орфографії (визначення відстані Левенштейна).
* Алгоритмів стиснення текстових даних та виявлення плагіату.
