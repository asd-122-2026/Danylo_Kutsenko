### Мета

## Хід роботи
### Завдання
Реалізувати алгоритм пошуку в ширину (BFS) для знаходження найкоротшого шляху у графі та візуалізувати процес обходу.

### Виконання
Граф — це сукупність вершин та ребер. Пошук у ширину (Breadth-First Search) дозволяє систематично обходити вершини графа, починаючи з початкової, рівень за рівнем.

```python
from collections import deque

def bfs(graph, start_node):
    visited = set()
    queue = deque([start_node])
    visited.add(start_node)
    
    order = []
    
    while queue:
        vertex = queue.popleft()
        order.append(vertex)
        
        for neighbour in graph[vertex]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)
    return order

# Приклад графа (словник суміжності)
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print("Порядок обходу BFS:", bfs(graph, 'A'))
```

## Висновок
У цій роботі я реалізував алгоритм пошуку в ширину. Він є ефективним для знаходження найкоротшого шляху у незважених графах. Часова складність алгоритму становить $O(V + E)$, де $V$ — кількість вершин, а $E$ — кількість ребер.

## Контрольні питання

### 1. Чим відрізняється BFS від DFS?
BFS (пошук у ширину) використовує чергу і рухається "шарами", тоді як DFS (пошук у глибину) використовує стек або рекурсію і йде максимально глибоко по одній гілці.

### 2. Яка структура даних найефективніша для BFS?
Черга (Queue), оскільки вона реалізує принцип FIFO (перший прийшов — перший пішов), що необхідно для обходу за рівнями.

### 3. Де застосовуються алгоритми на графах?
У навігаційних системах, соціальних мережах (пошук друзів), комп'ютерних мережах та для вирішення логічних задач.
