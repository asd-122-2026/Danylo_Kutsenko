## Хід роботи
### Завдання
Реалізувати структуру даних Стек (Stack) та Черга (Queue). Продемонструвати виконання основних операцій: додавання елемента, видалення елемента та перевірка стану структури.

### Виконання
Стек працює за принципом LIFO (Last In, First Out) — елемент, доданий останнім, видаляється першим. Черга працює за принципом FIFO (First In, First Out) — елемент, доданий першим, видаляється першим. Всі базові операції для цих структур виконуються за константний час O(1).

```python
from collections import deque

# --- 1. Реалізація Стека (Stack) ---
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)
        print(f"Стек: додано {item}")

    def pop(self):
        if not self.is_empty():
            item = self.items.pop()
            print(f"Стек: видалено {item}")
            return item
        print("Стек порожній!")

    def is_empty(self):
        return len(self.items) == 0

# --- 2. Реалізація Черги (Queue) ---
class Queue:
    def __init__(self):
        # deque оптимальніша для черги, ніж звичайний список
        self.items = deque()

    def enqueue(self, item):
        self.items.append(item)
        print(f"Черга: додано {item}")

    def dequeue(self):
        if not self.is_empty():
            item = self.items.popleft()
            print(f"Черга: видалено {item}")
            return item
        print("Черга порожня!")

    def is_empty(self):
        return len(self.items) == 0

# --- 3. Демонстрація роботи ---
print("--- ДЕМОНСТРАЦІЯ СТЕКА (LIFO) ---")
my_stack = Stack()
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)
my_stack.pop() # Має видалити 30
my_stack.pop() # Має видалити 20

print("\n--- ДЕМОНСТРАЦІЯ ЧЕРГИ (FIFO) ---")
my_queue = Queue()
my_queue.enqueue("Клієнт 1")
my_queue.enqueue("Клієнт 2")
my_queue.enqueue("Клієнт 3")
my_queue.dequeue() # Має видалити "Клієнт 1"
my_queue.dequeue() # Має видалити "Клієнт 2"
```

## Висновок
У ході лабораторної роботи було успішно реалізовано структури даних Стек та Черга. Експериментальна демонстрація підтвердила їхні теоретичні принципи: стек коректно обробляє дані за принципом LIFO, а черга — за принципом FIFO. Для оптимізації черги в Python було використано структуру `collections.deque`, яка дозволяє видаляти перший елемент за час O(1), уникаючи зсуву всього масиву.

## Контрольні питання

### 1. Які основні операції застосовуються для роботи зі стеком?
Основними операціями є `push` (додавання елемента на вершину стека) та `pop` (видалення і повернення елемента з вершини). Також часто реалізують `peek` (перегляд вершини без видалення).

### 2. Чому для реалізації черги в Python краще використовувати `deque`, а не `list`?
У звичайному списку `list` операція видалення першого елемента (`pop(0)`) вимагає зсуву всіх наступних елементів на одну позицію вліво, що займає час O(n). `deque` (двостороння черга) реалізована як двобічно зв'язний список, тому видалення з початку (`popleft()`) виконується миттєво за час O(1).

### 3. Де на практиці застосовуються ці структури?
- **Стек:** Реалізація механізму "Скасувати" (Undo) в редакторах, кнопка "Назад" у браузері, стек викликів функцій у програмуванні.
- **Черга:** Черга друку документів на принтер, буферизація відео або аудіо, алгоритми обходу графа в ширину (BFS).
